```python
# Lists

A built - in data types that stores set of values
It can store elements of different types (integer, float, string etc.)
```

```python
# Example

marks = [23, 45, 98, 32]
student = ["Tuhin", 21, "CSE(AIML)", "Howrah"]
```

```python
# Difference between Strings and Lists

Strings ---> immutable (can't change)
Lists ---> mutable (can change)
```

```python
marks = [23, 24, 25, 26]
print(marks)
print(len(marks))
print(marks[2])
```

```python
# Replace value is possible in Lists

student = ["Aakash", "Rahul", "Suman", "Raj"]
print(student)
student[0] = "Sanjib"
print(student)
```

```python
# List Slicing

student = ["Aakash", "Rahul", "Suman", "Raj"]
print(student[1:3])

# [:2] define as [0:2]

student = ["Aakash", "Rahul", "Suman", "Raj"]
print(student[:2])

# [1:] define as [1:len(str)]

student = ["Aakash", "Rahul", "Suman", "Raj"]
print(student[1:])

# Negative Indexing

student = ["Aakash", "Rahul", "Suman", "Raj"]
print(student[-3:-1])
```

```python
# Methods

# adds one element at the end

list.append(element)

# sorts in ascending order

list.sort()

# sorts in descending order

list.sort(reverse = True)

# reverse the list

list.reverse

# insert element at index

list.insert(index, element)
```

```python
list = [2, 1, 3]

list.append(4)
print(list)

list.sort()
print(list)

list.sort(reverse = True)
print(list)

list.reverse()
print(list)

list.insert(2, 8)
print(list)
```

```python
list = [2, 1, 3]
list.reverse()
print(list)
```

```python
# Remove first occurrence of element

list.remove(element)

# Remove element from index

list.pop(index)
```

```python
list = [1, 3, 1, 4]

list.remove(1)
print(list)

list.pop(2)
print(list)
```

```python
# Tuples

A built - in data types that lets us create immutable sequences of values
```

```python
# Difference between Lists and Tuples

Tuples ---> immutable (can't change)
Lists ---> mutable (can change)
```

```python
# Replace value is possible in Lists

student = ["Aakash", "Rahul", "Suman", "Raj"]
print(student)
student[0] = "Sanjib"
print(student)

# O/P:
['Aakash', 'Rahul', 'Suman', 'Raj']
['Sanjib', 'Rahul', 'Suman', 'Raj']

# Replace value is not possible in Tuples

student = ("Aakash", "Rahul", "Suman", "Raj")
print(student)
student[0] = "Sanjib"
print(student)

# O/P:
TypeError: 'tuple' object does not support item assignment
```

```python
# empty tuple is valid

tup = ()
print(tup)
print(type(tup))
```

```python
# If we don't use comma separated (,) then the type will be different. Type will be int, float, string

tup = (1)
print(tup)
print(type(tup))

tup = (1.0)
print(tup)
print(type(tup))

tup = ("Python")
print(tup)
print(type(tup))
```

```python
# If there is only one element is presente in the tuple, still i have to use comma (,).This comma(,) represent the Tuple

tup = (1,)
print(tup)
print(type(tup))
```

```python
# Slicing

tup = (1, 2, 3, 4)
print(tup[1:3])
```

```python
# Tuple Methods

# Returns index of first occurrence
tup.index(element)

# counts total occurrences
tup.count(element)
```

```python
tup = (1, 2, 3, 1, 2, 4 ,5)

print(tup.index(2))

print(tup.count(1))
```
