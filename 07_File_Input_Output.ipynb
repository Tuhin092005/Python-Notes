```python
# Open, read & close File

We have to open a file before reading or writing.

f = open(“file_name”, “mode”)

file name--> sample.txt / demo.docx
mode--> r : read mode / w : write mode

data = f.read( )
f.close( )
```

```python
Character          Meaning

'r'                open for reading (default)
'w'                open for writing, truncating the file first
'x'                create a new file and open it for writing
'a'                open for writing, appending to the end of the file if it exists
'b'                binary mode
't'                text mode (default)
'+'                open a disk file for updating (reading and writing)
```

```python
# with syntax

# doesn't require to write f.close(), it automatically close the file after performing the operation(S)

# with Syntax for reading

with open("demo.txt", "r") as f:
  data = f.read( )

# with Syntax for writing

with open("demo.txt", "w") as f:         # overwrites the entire file
  data = f.write("write something")

# with Syntax for appending

with open("demo.txt", "a") as f:         # adds to the file
  data = f.write("write something")
```

```python
# Search if the word "learning" exists in the file or not.

file = open("practice.txt", "r")
data = file.read()

if(data.find("learning") != -1):
  print("Found")
else:
  print("Not Found")
```

```python
# WAF to find in which line of the file does the word "learning" occur first.
# Print -1 if word not found.

file = open("practice.txt", "r")

line_no = 1

for line in file:
    if(line.find("learning") != -1):
        print(line_no)
        break
    line_no += 1
else:
  print(-1)

file.close()
```
