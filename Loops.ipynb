```python
# While Loops

# syntax
initialization
while condition:
  some work
  increment / decrement
```

```python
# Search for a number x in this tuple using loop:
#    (1, 4, 9, 16, 25, 36, 49, 64, 81, 100)

tup = (1, 4, 9, 16, 25, 36, 49, 64, 81, 100)

x = int(input("Enter a number to search: "))

i = 0

while(i < len(tup)):
  if(tup[i] == x):
    print(f"The number found at index:",i)
    break
  i = i+1
else:
    print("Not Found")
```

```python
# If same element present multiple times

# Search for a number x in this tuple using loop:
# (1, 4, 9, 16, 25, 36, 49, 64, 81, 100, 4)

tup = (1, 4, 9, 16, 25, 36, 49, 64, 81, 100, 4)

x = int(input("Enter a number to search: "))

i = 0

while(i < len(tup)):
  if(tup[i] == x):
    print(f"The number found at index:",i)
  else:
    print("Not Found")
  i = i+1
```

```python
# Break

used to terminate the loop when encountered.
```

```python
# Example of Break:

# Find the first occurrence of 4 in the given tuple

tup = (1, 4, 9, 16, 25, 36, 49, 64, 81, 100, 4)

x = int(input("Enter a number to search: "))

i = 0

while(i < len(tup)):
  if(tup[i] == x):
    print(f"The number found at index:",i)
    break
  i +=1
else:
  print("Not Found")
```

```python
# Continue

terminates execution in the current iteration & continues
execution of the loop with the next iteration.
```

```python
# Example of Continue:

# Print 1 to 10 without printing 6

for i in range(1, 11):
  if(i == 6):
    continue
  print(i)
```

```python
# print the even numbers from 1 to 11

for i in range(1, 11):
  if(i % 2 != 0):
    continue
  print(i)
```

```python
# print the odd numbers from 1 to 10

for i in range(1, 10):
  if(i % 2 == 0):
    continue
  print(i)
```

```python
# For Loops

Loops are used used for sequential traversal. For traversing list, string, tuples etc.
```

```python
# For loops:

# Syntax:

# for elements in list:
#   some work

list = [1, 2, 3, 4]

for i in list:
  print(i)
```

```python
# For loop with else:

# Syntax:

# for elements in list:
#   some work
# else:
#   works when loop ends

# use of else is optional, it is not mandatory
# It will be used when we use break statement

list = [1, 2, 3, 4]

for i in list:
  print(i*i)
else:
  print("stop")
```

```python
# Print the elements of the following list using a for loop
#       [1, 4, 9, 16, 25, 36, 49, 64, 81, 100]

nums = [1, 4, 9, 16, 25, 36, 49, 64, 81, 100]

for ele in nums:
  print(ele)
```

```python
# Search for a number x in this tuple using for loop
#    (1, 4, 9, 16, 25, 36, 49, 64, 81, 100, 4)

tup = (1, 4, 9, 16, 25, 36, 49, 64, 81, 100, 4)

x = int(input("Enter the number you want to search: "))

i = 0
for val in tup:
  if (val == x):
    print("found")
  i += 1
```

```python
# Search for a number x in this tuple using for loop
#    (1, 4, 9, 16, 25, 36, 49, 64, 81, 100, 4)

tup = (1, 4, 9, 16, 25, 36, 49, 64, 81, 100, 4)

x = int(input("Enter the number you want to search: "))

i = 0
for val in tup:
  if (val == x):
    print("found")
    break
  i += 1
else:
  print("not found")
```

```python
# Range()

Range functions returns a sequence of numbers, starting from 0 by default,
and increments by 1 (by default), and stops before a specified number.

# Syantax:

range(start?, stop, step?)

step size means --> increase
```

```python
# start = 0, stop = 5, step = 1
for i in range(5):
  print(i)

# start = 1, stop = 5, step = 1
for i in range(1, 5):
  print(i)

# start = 2, stop = 10, step = 2
for i in range(2, 10, 2):
  print(i)
```

```python
# Print numbers from 1 to 100
# using for and range()

for i in range(1, 101):
  print(i)
```

```python
# print numbers from 100 to 1
# using for and range()

for i in range(100, 0, -1):
  print(i)
```

```python
# print the multiplication table of a number n.

n = int(input("Enter the value for n: "))

for i in range(1, 11):
  print(n * i)
```

```python
# WAP to find the sum of first n numbers. (using while)

num = int(input("Enter the value for n: "))

i = 1
sum = 0

while(i <= num):
  sum = sum+i
  i = i+1

print("Total Sum:",sum)
```

```python
# WAP to find the factorial of first n numbers. (using for)

n = int(input("Enter a no.: "))

i = 1
fact = 1

for i in range(1, n+1):
  fact = fact*i

print(f"Factorial of {n} is:", fact)
```
