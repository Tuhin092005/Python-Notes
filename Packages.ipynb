```python
# 1. What is a Package ?

• A package is a way of organizing related modules in a directory.

• It contains a special file called __init__.py

• Packages help in code reusability and scalability.

Example Package Structure

mypackage/
├── __init__.py
├── math_utils.py
├── string_utils.py
main.py

★ mypackage is a package.

★ It contains modules.
```

```python
# 2. Creating a Package

• Create a folder with the package name.

• Inside it, add a file named __init__.py (can be empty).

• Add your module files in the same folder.

Example :

mkdir mypackage
cd mypackage

touch __init__.py      # create empty file
touch math_utils.py
touch string_utils.py
```

```python
# 3. Importing from a Package

• We can import modules or specific items from a package.

(a) Import Module

from mypackage import math_utils

result = math_utils.add(5, 3)
print(result)

(b) Import Specific Item

from mypackage.math_utils import add

result = add(5, 3)
print(result)

★ Always use dot (.) operator to access modules and functions.
```

```python
# 4. The __init__.py File

• This file is executed when the package is imported.

• We can use it to initialize the package.

• It can also contain package-level variables or import statements.

Example : __init__.py

print("Initializing mypackage...")

__version__ = "1.0.0"

from .math_utils import add

★ Executed when package is imported.
```

```python
# 5. Absolute vs Relative Import

       Absolute Import	                    Relative Import
from mypackage import math_utils	     from . import math_utils
Starts from top-level package.	       Starts from current package.
Works clearly and is preferred.	       Used inside packages.

★ Use absolute imports in main programs.

★ Relative imports are useful inside packages.
```

```python
# 6. Built-in / Standard Packages

• Python comes with many built-in packages in the standard library.

• Some useful ones :

Package	    Use
math	      Mathematical functions
random	    Generate random numbers
datetime	  Work with date & time
os	        Operating system interaction
sys	        System-specific parameters
json	      Work with JSON data

★ No need to install these, they are ready to use!
```

```python
# 7. Installing External Packages

• External packages are not included in Python.

• Install using pip.

• Install from Python Package Index (PyPI)

pip install package_name            # install latest version

pip install package_name == 1.2.0   # specific version

Example :

pip install requests

pip install numpy == 1.24.0
```

```python
# 8. Listing Installed Packages

• See all installed packages using pip.

pip list      # list installed packages

pip freeze    # list with versions

Uninstall a package :

pip uninstall package_name
```
