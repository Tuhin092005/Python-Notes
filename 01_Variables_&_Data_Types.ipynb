```python
# Types of Operators

# Definition

An operator is a symbol that performs an operation between operands.
```

```python
# 1. Arithmetic Operators

Operator    Meaning
+	          Addition
-	          Subtraction
*	          Multiplication
/	          Division
%	          Modulus
**	        Exponent
```

```python
# Example

a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b)
print(a ** b)
```

```python
# 2. Relational / Comparison Operators

Operator    Meaning
==	        Equal
!=	        Not Equal
>	          Greater Than
<	          Less Than
>=	        Greater Than or Equal
<=	        Less Than or Equal
```

```python
# Example

a = 10
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

```python
# 3. Assignment Operators

Operator    Example
=	          a = 5
+=	        a += 2
-=	        a -= 2
*=	        a *= 2
/=	        a /= 2
%=	        a %= 2
**=	        a **= 2
```

```python
# Example

a = 5

a += 2
print(a)

a *= 3
print(a)
```

```python
# 4. Logical Operators

Operator    Meaning
and	        Both conditions must be True
or	        At least one condition is True
not	        Reverses the result
```

```python
# Example

a = True
b = False

print(a and b)
print(a or b)
print(not a)
```

```python
# Type Conversion

Python automatically converts compatible data types.

# Example

a = 1
b = 2.0

sum = a + b
print(sum)
print(type(sum))

# Output:

3.0
<class 'float'>

# Invalid Conversion

a = 1
b = "2"

sum = a + b   # Error

This produces a TypeError because an integer and a string cannot be added directly.
```

```python
# Type Casting

Type casting means manually converting one data type into another.

# Example

a = 1
b = "2"

c = int(b)
sum = a + c

print(sum)

# Output:

3

# More examples:

print(float(5))
print(str(100))
print(bool(1))
```
