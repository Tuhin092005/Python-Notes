```python
# What is a Module?

• A module is a file containing Python code (functions, classes, variables)

• We can import and reuse code from other modules.

• Every Python file is a module.

mymath.py
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

We can import this module in another file.

main.py
import mymath

result = mymath.add(5, 3)

print(result)
```

```python
# 2. Creating a Module

• Create a Python file with .py extension.

• Write functions, classes, variables in it.

Example : mymath.py

def add(a, b):
    return a + b

def mul(a, b):
    return a * b

PI = 3.14159

★ This file is now a module named mymath
```

```python
# 3. Importing Modules

• We use import statement.

• Two ways to import :

(a) Import the whole module

import mymath

print(mymath.add(2, 3))

print(mymath.PI)

(b) Import specific items

from mymath import add, mul

print(add(2, 3))

print(mul(2, 3))

★ Using specific import makes code shorter and easier to read.
```

```python
# 4. Using ‘as’ Keyword

• We can use as to give alias to module.

import mymath as mm

print(mm.add(4, 5))

print(mm.PI)

★ mm is an alias for mymath
```

```python
# 5. Built-in Modules

• Python has many built-in modules.

• Some commonly used modules :

Module	    Purpose
math	      Mathematical functions
random	    Generate random numbers
datetime	  Work with dates and times
os	        Interact with operating system
sys	        System-specific parameters

★ No need to create these modules, they are ready to use!
```

```python
# 6. Example – Using Built-in Module

• Using math module

import math

print(math.sqrt(16))     # 4.0

print(math.pi)           # 3.141592653589793
```

```python
# 7. The __name__ Variable

• Every Python module has a special variable name.

• When a module is run directly, name is set to "main".

• Otherwise, it is set to the module name.

Example : mymodule.py

def greet():
    print("Hello from module!")

if __name__ == "__main__":
    greet()     # Runs only if file is executed directly

★ Good for testing code in a module.
```
