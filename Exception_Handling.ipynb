```python
# 1. What is Exception?

• An exception is an error that occurs during the execution of a program.

• Python provides a way to handle these errors so the program doesn't crash.
```

```python
# 2. Basic Syntax

try:
  # code that might raise exception

except ExceptionType:
  # handle the exception

else:
  # runs if no exception occurs

finally:
  # always runs
```

```python
# 3. Simple Example

• finally block runs always!

try:
    num = int(input("Enter a number:"))
    result = 10 / num

except ZeroDivisionError:
    print("Cannot divide by zero!")

else:
    print("Result = ", result)

finally:
    print("Execution completed.")
```

```python
# 4. Else and Finally in Action

• else runs when if try block is successful.

• finally runs always!

try:
    x = int(input("Enter a number: "))
    print("Number is", x)

except ValueError:
    print("Invalid input!")

else:
    print("No exception occurred.")

finally:
    print("This will always run.")
```

```python
# 5. Multiple Except Blocks

try:
    a = int(input("Enter numerator: "))
    b = int(input("Enter denominator: "))
    print("Result = ", a / b)

except ZeroDivisionError:
    print("Denominator cannot be zero!")

except ValueError:
    print("Please enter valid numbers!")

except Exception as e:
    print("Some other error occurred:", e)
```

```python
# 6. Common Built-in Exceptions

Exception                Description
ZeroDivisionError        Raised when division by zero occurs
ValueError               Raised when a function gets an argument of correct type but inappropriate value
TypeError                Raised when an operation or function is applied to an inappropriate type
FileNotFoundError        Raised when a file is not found
IndexError               Raised when index is out of range
KeyError                 Raised when key is not found in dictionary
```

```python
# 7. Raising Exceptions

• We can raise exceptions manually using raise.

• Useful to enforce conditions in our code.

def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or above!")
    print("Access Granted")

check_age(16)       # Raises ValueError
```

```python
# 8. Custom Exception

• We can create our own exceptions by creating a new class.

• It should be inherited from Exception class.

class InvalidAgeError(Exception):
    pass

def validate_age(age):
    if age < 0 or age > 120:
        raise InvalidAgeError("Invalid age!")

try:
    validate_age(200)

except InvalidAgeError as e:
    print("Custom Exception:", e)
```
