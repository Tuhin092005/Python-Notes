```python
# 1. What is a Virtual Environment?

• A virtual environment (venv) is an isolated Python environment.

• It has its own Python interpreter and packages.

• Helps avoid dependency conflicts between projects.

• Keeps global Python installation clean.

★ Think of it as a separate workspace for your project.
```

```python
# 2. Why Use Virtual Environments?

• Different projects may need different versions of the same package.

• Prevents "it works on my machine" problems.

• Easy to share project requirements.

• Keeps system Python clean & safe.

• Good practice for professional development.
```

```python
# 3. Creating a Virtual Environment

• Python comes with built-in venv module.

• Create a new virtual environment using:

# Windows

python -m venv myenv

# macOS / Linux

python3 -m venv myenv

★ This creates a folder "myenv" with its own Python and packages.
```

```python
# 4. Activating the Virtual Environment

• Activate it before using.

# Windows (Command Prompt)

myenv\Scripts\activate

# macOS / Linux (Terminal)

source myenv/bin/activate

★ After activation, your prompt will look like:

(myenv) C:\Users\Tuhin>
(myenv) tuhin@mac ~ $
```

```python
# 5. Deactivating the Environment

• When done, deactivate it:

deactivate

★ Works on all platforms.
```

```python
# 6. Installing Packages in the Environment

• All packages installed now will be inside this environment.

• Use pip as usual.

pip install numpy
pip install requests

★ Installed only in this virtual environment.
```

```python
# 7. Viewing Installed Packages

• List all packages in the current environment.

pip list

★ Shows all installed packages.

Or with details:

pip freeze

★ Shows exact versions. Useful for sharing.
```

```python
# 8. Saving Dependencies

• Save installed packages to a requirements file.

pip freeze > requirements.txt

★ Creates a file with all packages and versions.
```

```python
# 9. Installing from Requirements File

• Recreate the same environment on another machine.

pip install -r requirements.txt

★ Installs all packages listed in the file.
```
