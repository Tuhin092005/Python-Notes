```python
# 1. PYTHON OOPS – INTRODUCTION

# What is OOPS?

OOPS stands for Object-Oriented Programming System.

Python is an object-oriented programming language where everything is treated as an object.

# Advantages of OOPS

a. Code reusability
b. Better organization
c. Easy maintenance
d. Data security
e. Modularity

# Main OOPS Concepts

a. Class
b. Object
c. Inheritance
d. Polymorphism
e. Encapsulation
f. Abstraction
```

```python
# 2. CLASS AND OBJECT

# Class

A class is a blueprint or template used to create objects.

# Object

An object is an instance of a class.

A class can contain:

a. Attributes (variables)
b. Methods (functions)

# Example: Class and Object

class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def show(self):
        print(self.brand, self.color)

c1 = Car("BMW", "Black")
c2 = Car("Audi", "White")

c1.show()
c2.show()

# Output

BMW Black
Audi White

# Important Point

a. Car → class (blueprint)
b. c1, c2 → objects created from the class
```

```python
# 3. CONSTRUCTOR

# What is a Constructor?

A constructor is a special method that is automatically called when an object is created.

In Python, the constructor is:

__init__()

# Uses

a. Initialize object attributes
b. Assign initial values

# Example: Constructor

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Aman", 22)

print(p1.name)
print(p1.age)

# Output

Aman
22

# How it Works

Person("Aman", 22)
        ↓
__init__("Aman", 22)
        ↓
Attributes are initialized
```

```python
# 4. INHERITANCE

# Definition

Inheritance allows one class (child class) to acquire the properties and methods of another class (parent class).

# Benefits

a. Code reusability
b. Easy extension of existing code
c. Represents is-a relationship

# Types of Inheritance in Python

a. Single inheritance
b. Multiple inheritance
c. Multilevel inheritance
d. Hierarchical inheritance

# Example: Single Inheritance

class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

d = Dog()

d.speak()
d.bark()

# Output

Animal speaks
Dog barks

# Key Point

The child class can:

a. Use parent class methods
b. Have its own methods
```

```python
# 5. POLYMORPHISM

# Definition

Polymorphism means many forms.

The same method name can perform different actions depending on the object.

Python mainly supports polymorphism through method overriding.

# Example: Method Overriding

class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

a = Animal()
d = Dog()

a.speak()
d.speak()

# Output

Animal speaks
Dog barks

# Key Point

The method in the child class overrides the method in the parent class.
```

```python
# 6. ENCAPSULATION

# Definition

Encapsulation is the process of wrapping data and methods into a single unit (class).

It is used to restrict direct access to data.

# Example: Encapsulation

class Bank:
    def __init__(self):
        self.__balance = 1000

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

b = Bank()

b.deposit(500)

print(b.get_balance())

# Output

1500

# Invalid Access

print(b.__balance)

# Output

AttributeError

# Access Specifiers in Python

Specifier	    Syntax	    Access
Public	      member	    Anywhere
Protected	    _member	    Within class and subclasses
Private	      __member	  Only within the class

# Key Point

Encapsulation provides data hiding and protects data from unauthorized access.
```

```python
# 7. ABSTRACTION

# Definition

Abstraction hides the internal implementation details and shows only the essential features.

In Python

Abstraction is implemented using:

a. Abstract classes
b. Abstract methods

# Example: Abstraction

from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):

    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r * self.r

c = Circle(5)

print(c.area())

# Output

78.5

# Key Point

Abstraction focuses on:

a. What to do
b. not How to do it
```

```python
# 8. SPECIAL / MAGIC METHODS

# Definition

Python provides special methods (also called dunder methods) that begin and end with double underscores.

They customize the behavior of objects.

# Common Magic Methods

Method	         Purpose
__init__()	     Constructor
__str__()	       String representation
__len__()	       Length of object
__add__()	       Addition using +
__eq__()	       Equality comparison

# Example: Magic Methods

class Book:

    def __init__(self, title, price):
        self.title = title
        self.price = price

    def __str__(self):
        return f"{self.title} - {self.price}"

    def __add__(self, other):
        return self.price + other.price

b1 = Book("Python Basics", 300)
b2 = Book("Java Guide", 200)

print(b1)
print(b1 + b2)

# Output

Python Basics - 300
500
```
