```python
# Dictionaries are used to store data values in key:value pairs
# They are unordered, mutable (changeable) & don't allow duplicate keys

# Example
dict = {
    "name" : "Tuhin",
    "roll" : 61,
    "marks" : [78, 65, 97]
}

print(dict)
print(type(dict))

print(dict["name"])
print(dict["roll"])

# Change the name to Rahul

dict["name"] = "Rahul"
print(dict)

# Also can add new value

dict["age"] = 21
print(dict)
```

```python
# Empty dictionary

null_dict = {

}
print(null_dict)
```

```python
# Nested Dictionaries

student = {
    "name" : "Tuhin",
    "marks" : {
        "phys" : 75,
        "chem" : 61,
        "math" : 88
    }
}

print(student)
print(student["marks"]["chem"])
```

```python
# Total number of key value pair

print(len(student))
```

```python
# Dictionary Methods

# returns all keys
myDict.keys()

# returns all values
myDict.values()

# returns all (key, value) pairs as tuples
myDict.items()

# returns the key according to value
myDict.get("Key")

# inserts the specified items to the dictionary
```

```python
print(student.keys())

print(student.values())

print(student.items())

print(student.get("name"))
print(student.get("marks").get("phys"))

student.update({"city" : "Kolkata", "age" : 21})
print(student)
```

```python
# Set

Set is collection of unordered items.

Each element in the set must be unique & immutable.

Within Set, we cannot store List & Dict, as they both are mutable.
```

```python
nums = {1, 2, 3, 4}
my_set = {1, 2, 2, 2}

# Repeated items stored only once, so it resolved to {1, 2}
```

```python
# NULL Set / empty set

null_set = set()
print(null_set)
```

```python
set1 = {1, 2, 3, 4}
print(set1)
print(type(set1))
```

```python
# Duplicate entries will be resolved to single entries

set2 = {1, 2, 2, 3, 4, "Tuhin", "Rahul", "Rahul"}
print(set2)
```

```python
# Set Methods

# adds an element
set.add(element)

# removes an element
set.remove(element)

# empties the set
set.clear()

# removes a random variable
set.pop()
```

```python
# Sets --> mutable
# Set --> elements --> immutable
```

```python
collection = {1, 2, 3, 3, 4, 5, 5}

collection.add(6)
print(collection)

collection.remove(1)
print(collection)

collection.clear()
print(collection)

collection1 = {1, 2, 2, 3, 4, 4, 5}
collection1.pop()
collection1.pop()
print(collection1)
print(collection1)
```

```python
# More Set Methods

# combines both set values and returns new
set.union(set2)

# combines common values and returns new
set.intersection(set2)
```

```python
# Example

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

union = a.union(b)
print(union)
```

```python
# Example

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

intersection = a.intersection(b)
print(intersection)
```
