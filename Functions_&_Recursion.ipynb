```python
# Functions

Block of statements that perform a specific task.

# Syntax:

def func_name( param1, param2..):     # Function Definition
  #some work
  return val

func_name( arg1, arg2 ..)   # Function Call
```

```python
# Function Definition
def sum(a, b):  #a, b --> parameters
  s = a + b
  return s

# Function Call
sum(2, 3)    #2, 3 --> arguments
```

```python
# Default Parameters

Assigning a default value to parameter, which is used when no argument is passed.
```

```python
# Example of Default Parameters

def sum(a = 1, b = 2):
  return a + b

sum()
```

```python
# WAF to print the length of a list. (list is the parameter)

list1 = [1, 2, 3, 4, 5]

def cal_length(list):
  return len(list)

cal_length(list1)
```

```python
# WAF to print the elements of a list in a single line. (list is the parameter)

list2 = ["a", "b", "c", "d", "e"]

def print_list(list):
  for i in list:
    print(i, end = " ")

print_list(list2)
```

```python
# WAF to find the factorial of n. (n is the parameter)

def cal_fact(n):
  fact = 1
  for i in range(1, n+1):
    fact = fact * i
  return fact

cal_fact(5)
```

```python
# WAF to convert USD to INR.

def convert(usd_value):
  inr_value = usd_value * 96.86
  return inr_value

convert(5)
```

```python
# WAF to check a number is ODD or EVEN

def odd_even(num):
  if(num % 2 == 0):
    print("Even")
  else:
    print("Odd")

odd_even(5)
odd_even(4)
```

```python
# Recursion

When a function calls itself repeatedly.
```

```python
# Example of Recursive function

# print 5 4 3 2 1

def show(n):
  if(n == 0):  # base case
    return 0
  print (n)
  show (n-1)

show(5)

print("\n")

# print 1 2 3 4 5

def show(n):
  if(n == 6):  # base case
    return 0
  print (n)
  show (n+1)

show(1)
```

```python
# Factorial by using Recursive Function

def fact(n):
  if(n == 0 or n == 1):
    return 1
  else:
    return n * fact(n - 1)

fact(4)
```

```python
# Write a recursive function to calculate the sum of first n natural numbers.

def cal_sum(n):
  if( n == 0):
    return 0
  else:
    return n + cal_sum(n - 1)

cal_sum(5)
```

```python
# Write a recursive function to print all elements in a list.

num = [1, 2, 3, 4, 5]

def show(list, i):
  if(i == len(list)):
    return 0
  print(list[i])
  show(list, i+1)

show(num, 0)
```

```python
# Write a recursive function to print all elements in a list.

list = [1, 2, 3, 4, 5]

def show(list):
  if(len(list) == 0):
    return 0
  print(list[0])
  show (list[1:len(list)])

show(list)
```
